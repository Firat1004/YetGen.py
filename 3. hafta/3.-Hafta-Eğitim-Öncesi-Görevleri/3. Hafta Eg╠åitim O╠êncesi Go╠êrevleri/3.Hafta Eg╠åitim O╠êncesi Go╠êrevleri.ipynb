# Problem 1
Bir sayının 7'ye bölünebilir olup olmadığını kontrol etmek 
için bir program yazın.

İpucu: Mod operatörünü (%) kullanabilirsiniz.

````python
sayi = int(input("kontrol etmek istediğiniz sayıyı girin"))
bolen = 7
if sayi != 0 :
    if sayi % bolen == 0:
        print("sayınız 7'ye tam bölünmektedir")
    else:
        print("sayınız 7'ye tam bölünmemektedir")
        
elif sayi == 0:
    print("0 ile bölme işlemi yapılamaz")

else :
    print("hata")
````

# Problem 2
Kullanıcıdan vize ve final notu isteyin. Girilen vize notunun %40’ı ve 
girilen finalnotunun ise %60’ı alınarak yıl sonu not ortalaması hesaplanacaktır. 

Bu not ortalaması eğer 85 ve üzeri ise AA, 

75 ve 85 arasında ise BA, 

70 ve 75 arasında ise BB, 

65 ve 70 arasında ise CB,

60 ve 65 arasında ise CC,

55 ve 60 arasında ise DC,

50 ve 55 arasında ise DD olarak hesaplanacaktır. 

Bu öğrencinin yıl sonu toplam notu 50’nin altında ise FF ile dersten kalacaktır. 

Ayrıca öğrencinin final notu 50’nin altında ise direkt FF ile kalacaktır.

````python
vize = int(input("vize notunu giriniz"))
final = int(input("final notunu giriniz"))

sonuc = vize * 0.4 + final * 0.6

if vize <= 100 | final <= 100:
     
    if final > 49:

        if sonuc >= 85:
            print("AA")
        elif sonuc > 74 & sonuc < 85:
            print("BA")

        elif sonuc > 69 & sonuc < 75:
            print("BB")
            
        elif sonuc > 64  & sonuc < 70:
            print("CB")

        elif sonuc > 59 & sonuc < 65:
            print("CC")

        elif sonuc > 54 & sonuc <60:
            print("DC")

        elif sonuc > 49 & sonuc < 55:
            print("DD")

        elif sonuc < 50:
            print("FF")

    elif final < 50 :
        print("final notunuz 50'den düşük olduğu için kaldınız")

elif vize > 100 | final > 100:
    print("hatalı not girdisi")
````

```
 BKİ 18.5'un altındaysa -------> Zayıf
```
