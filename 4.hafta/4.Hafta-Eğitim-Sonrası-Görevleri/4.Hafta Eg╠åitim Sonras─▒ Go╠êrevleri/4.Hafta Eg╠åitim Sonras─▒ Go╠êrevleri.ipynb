### 🌶️ b) Hadi işi biraz daha zorlaştıralım! 


Şimdiyse senden `numbers` isimli listedeki tekrar eden elemanları `repeatNumbers` adındaki listeye; tekrar etmeyen elemanları `uniqueNumbers`adındaki listeye kopyalayan yazmanı ve listeleri ekrana bastırmanı bekliyoruz

```
>Output

```

````python
# 2.Problemin b Şıkkının Çözümünü Buraya Yazınız
from collections import Counter
numbers =[5,10,10,15,15,20,20,20,]

say = Counter(numbers)

tekrar_edenler = [num for num, count in say.items() if count >1]
tekrar_etmeyenler = [num for num, count in say.items() if count == 1]

print(f"tekrar edenler {tekrar_edenler}")
print(f"tekrar etmeyenler {tekrar_etmeyenler}")


````

# Problem 4

Buradaki işlemlerimizin tamamı stringler içindir. Aşağıda verilecek olan metodlar için kendinizin örnek yaratması gerekmektedir. 

Örneğin `upper` için `string` değişkeni oluştururken` k = "BAC"` şeklinde oluşturmayınız. Bunun yerine k değişkenini `k = "bac`" şeklinde oluşturmanız `upper` metodundaki değişimi görebilmeniz için daha iyi olacaktır.

Bütün metodlardaki değişimin gözle görülür şekilde değişimi beklenmektedir.


1- capitalize(), lower(), upper(), swapcase(), title(), replace()

2- endswith(), startswith(), find(), index(), count()

3- split(), strip(), join(), islower(), isalpha(), isdigit()


Yukarıdaki `string` metodlarındaki değişimleri görebildiğimiz bir program yazınız.


````python
# 4.Problemin Çözümünü Buraya Yazınız

k = "merhaba dünya"
print(k.capitalize())  
print(k.lower())  
print(k.upper())  
print(k.swapcase())  
print(k.title())  
print(k.replace("dünya", "evren"))  

k = "merhaba dünya"
print(k.endswith("dünya"))  
print(k.startswith("mer"))  
print(k.find("dünya"))  
print(k.index("dünya"))  
print(k.count("a"))  

k = "   merhaba dünya   "
print(k.split())  
print(k.strip())  
print("-".join(["merhaba", "dünya"]))  
print(k.islower())  
print(k.isalpha())  
print("123".isdigit())  

````
