# Problem 1

Kumanda isminde bir sınıf oluşturarak aşağıdaki işlemleri yapmasını sağlayan fonksiyonları sınıf içerisinde yazalım.
1. Tv Aç
2. Tv Kapat
3. Ses Ayarları
4. Kanal Ekle
5. Kanal Sayısını Öğrenme
6. Rastgele Kanala Geçme
7. Televizyon Bilgileri


Kumandayı kapatmak için q tuşuna basılmalı ve diğer durumlarda hangi işlemi yapmak istediği sorulmalıdır. Her işlemde gerekli fonksiyonunun sınıf içerisinden çağırılması gerekmektedir.

random ve time kütüphanesini araştırarak kullanabilirsiniz.

Kendi istediğiniz özellikleri eklemekte serbetsiniz.

```python
# 1.Problemin Çözümünü Buraya Yazınız
import random
import time

class Kumanda:
    def __init__(self):
        self.tv_acik = False
        self.ses = 10
        self.kanallar = ["Kanal 1", "Kanal 2", "Kanal 3", "Kanal 4"]
        self.current_channel = self.kanallar[0]

    def tv_ac(self):
        self.tv_acik = True
        print("Televizyon açıldı.")

    def tv_kapat(self):
        self.tv_acik = False
        print("Televizyon kapatıldı.")

    def ses_ayarları(self):
        while True:
            try:
                yeni_ses = int(input("Ses seviyesini ayarlayın (0-100): "))
                if 0 <= yeni_ses <= 100:
                    self.ses = yeni_ses
                    print(f"Ses seviyesi {self.ses}.")
                    break
                else:
                    print("Ses seviyesi 0 ile 100 arasında olmalı.")
            except ValueError:
                print("Lütfen geçerli bir sayı girin.")

    def kanal_ekle(self):
        kanal = input("Yeni kanal adı girin: ")
        self.kanallar.append(kanal)
        print(f"{kanal} kanalı eklendi.")

    def kanal_sayisi(self):
        print(f"Toplam kanal sayısı: {len(self.kanallar)}")

    def rastgele_kanala_gec(self):
        kanal = random.choice(self.kanallar)
        self.current_channel = kanal
        print(f"Rastgele kanala geçildi: {kanal}")

    def televizyon_bilgileri(self):
        acik_durum = "Açık" if self.tv_acik else "Kapalı"
        print(f"Televizyon Durumu: {acik_durum}")
        print(f"Ses Seviyesi: {self.ses}")
        print(f"Şu Anki Kanal: {self.current_channel}")

    def kumanda_ekrani(self):
        while True:
            print("\nYapmak istediğiniz işlemi seçin:")
            print("1. Tv Aç")
            print("2. Tv Kapat")
            print("3. Ses Ayarları")
            print("4. Kanal Ekle")
            print("5. Kanal Sayısını Öğrenme")
            print("6. Rastgele Kanala Geçme")
            print("7. Televizyon Bilgileri")
            print("q. Çıkış")
            
            secim = input("Seçiminiz: ").lower()
            
            if secim == "1":
                self.tv_ac()
            elif secim == "2":
                self.tv_kapat()
            elif secim == "3":
                self.ses_ayarları()
            elif secim == "4":
                self.kanal_ekle()
            elif secim == "5":
                self.kanal_sayisi()
            elif secim == "6":
                self.rastgele_kanala_gec()
            elif secim == "7":
                self.televizyon_bilgileri()
            elif secim == "q":
                print("Kumanda kapatılıyor.")
                break
            else:
                print("Geçersiz seçim.")

kumanda = Kumanda()
kumanda.kumanda_ekrani()
```
